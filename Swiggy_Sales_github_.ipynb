###**CHARTS DESIGN**

Convert Date Column to Datetime

```python
# Step 1 — Date Conversion

df['Order Date'] = pd.to_datetime(df['Order Date'])
```

```python
# Step 2 — Create Time Columns

df['month'] = df['Order Date'].dt.month_name()
df['day'] = df['Order Date'].dt.day_name()
df['quarter'] = df['Order Date'].dt.quarter
```

Monthly Sales Trend

```python
# Step 3 — Monthly Sales Trend

monthly_sales = df.groupby('month')['Price (INR)'].sum()

plt.figure(figsize=(10,5))
monthly_sales.plot(kind='line', marker='o')

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales (INR)")

plt.show()
```

Daily Sales Trend

```python
# Step 4 — Daily Sales Trend

daily_sales = df.groupby('day')['Price (INR)'].sum()

plt.figure(figsize=(10,5))
daily_sales.plot(kind='bar')

plt.title("Daily Sales Trend")
plt.xlabel("Day")
plt.ylabel("Total Sales (INR)")

plt.show()
```

Total Sales by Food Type (Veg vs Non-Veg)

```python
# Step 5A — Create Food Type Column

df['food_type'] = df['Dish Name'].apply(
    lambda x: 'Veg' if 'veg' in str(x).lower() else 'Non-Veg'
)
```

```python
# Step 5B — Sales by Food Type

food_sales = df.groupby('food_type')['Price (INR)'].sum()
print(food_sales)
```

```python
# Step 5C — Visualization

plt.figure(figsize=(6,6))

plt.pie(food_sales,
        labels=food_sales.index,
        autopct='%1.1f%%',
        startangle=90)

plt.title("Total Sales by Food Type (Veg vs Non-Veg)")

plt.show()
```

Total Sales by State

```python
# Step 6 — Sales by State

state_sales = df.groupby('State')['Price (INR)'].sum()

plt.figure(figsize=(10,5))
state_sales.plot(kind='bar')

plt.title("Total Sales by State")
plt.xlabel("State")
plt.ylabel("Total Sales (INR)")

plt.show()
```

Quarterly Performance Summary

```python
# Step 7 — Quarterly Sales Summary

quarter_sales = df.groupby('quarter')['Price (INR)'].sum()

plt.figure(figsize=(8,5))
quarter_sales.plot(kind='bar')

plt.title("Quarterly Performance Summary")
plt.xlabel("Quarter")
plt.ylabel("Total Sales (INR)")

plt.show()
```

Top 5 Cities by Sales

```python
# Step 8 — Top 5 Cities by Sales

top_cities = df.groupby('City')['Price (INR)'].sum().sort_values(ascending=False).head(5)

plt.figure(figsize=(8,5))
top_cities.plot(kind='bar')

plt.title("Top 5 Cities by Sales")
plt.xlabel("City")
plt.ylabel("Total Sales (INR)")

plt.show()
```

Rating Distribution Chart

```python
# Step 9 — Rating Distribution Chart

plt.figure(figsize=(8,5))
sns.histplot(df['Rating'], bins=20)
plt.title("Rating Distribution")
plt.xlabel("Rating")
plt.ylabel("Count")
plt.show()
```

Top Cuisines

```python
# Step 10 — Top Cuisines

top_cuisines = df['Category'].value_counts().head(10)

plt.figure(figsize=(8,5))
top_cuisines.plot(kind='bar')
plt.title("Top 10 Cuisines")
plt.xlabel("Cuisine")
plt.ylabel("Count")
plt.show()
```

Price vs Rating Analysis

```python
# Step 11 — Price vs Rating Analysis

plt.figure(figsize=(8,5))
sns.scatterplot(x='Price (INR)', y='Rating', data=df)

plt.title("Price vs Rating of Dishes")
plt.xlabel("Price (INR)")
plt.ylabel("Rating")

plt.show()
```

Most Popular Dishes

```python
# Step 12 — Most Popular Dishes

top_dishes = df['Dish Name'].value_counts().head(10)

plt.figure(figsize=(8,5))
top_dishes.plot(kind='bar')
plt.title("Top Ordered Dishes")
plt.show()
```

##**Key Business Insights**

**Top Revenue Drivers**:
Top-rated restaurants (4.5+ ⭐) generate a large share of total revenue.

**Recommendation**: Introduce a Premium Partner program to boost visibility and encourage exclusive menu offerings.

**Cuisine Demand**:
North Indian and Chinese cuisines dominate order volume, while Healthy/Salad options have high ratings but low availability.

**Recommendation**: Encourage restaurants to expand healthy food options in high-demand areas.

**Peak Ordering Hours**:
Most orders occur between 7 PM – 10 PM, especially on weekends.
**Recommendation**: Offer flash deals during off-peak hours (3 PM – 6 PM) to increase orders.

**Customer Loyalty**:
Frequent users prefer Free Delivery and Combo Deals.

**Recommendation**: Enhance the Swiggy One subscription with milestone rewards.

**Operational Efficiency**:
High restaurant density cities show lower average order value (AOV).

**Recommendation**: Use cross-selling and bundle offers to increase order value.
